```python
# Task 1: Student Grade Analyzer (Functions & Dictionaries)
#
# Write a Python function named analyze_student_grades(student_data).
# It accepts a dictionary where keys are student names
# and values are lists of their scores.
#
# 1. Calculate the average score for each student.
# 2. Return a new dictionary containing only students
#    who passed (average >= 60), using a comprehension or loop.
#
# Test Data:
# students_input = {
#     "Ahmed": [55, 60, 65],
#     "Sara": [85, 90, 95],
#     "Omar": [40, 45, 50],
#     "Mona": [70, 75, 80]
# }
#
# Expected Output:
# Passed Students Averages:
# {'Ahmed': 60.0, 'Sara': 90.0, 'Mona': 75.0}
```

```python
def analyze_student_grades(student_data):
    passed_students = {}

    for name in student_data:
        scores = student_data[name]
        average = sum(scores) / len(scores)

        if average >= 60:
            passed_students[name] = average

    return passed_students


students_input = {
    "Ahmed": [55, 60, 65],
    "Sara": [85, 90, 95],
    "Omar": [40, 45, 50],
    "Mona": [70, 75, 80]
}

print("Passed Students Averages:")
print(analyze_student_grades(students_input))
```

```python
# Task 2: Inventory Price Update
# (Dictionary Comprehensions & Conditional Logic)
#
# Given a dictionary of store items and their prices:
# 1. Filter out items with a price less than or equal to 500.
# 2. Apply a 15% discount (price * 0.85) to the remaining items.
# Use a Dictionary Comprehension.
#
# Test Data:
# store_inventory = {
#     "Laptop": 15000,
#     "Mouse": 250,
#     "Keyboard": 600,
#     "USB Cable": 150,
#     "Monitor": 4000
# }
#
# Expected Output:
# Discounted Premium Items:
# {'Laptop': 12750.0, 'Keyboard': 510.0, 'Monitor': 3400.0}
```

```python
store_inventory = {
    "Laptop": 15000,
    "Mouse": 250,
    "Keyboard": 600,
    "USB Cable": 150,
    "Monitor": 4000
}

discounted_items = {
    item: price * 0.85
    for item, price in store_inventory.items()
    if price > 500
}

print("Discounted Premium Items:")
print(discounted_items)
```

```python
# Task 3: NumPy Array Reshaping & Slicing
#
# Problem Description:
# 1. Set the random seed to 42.
# 2. Generate a 1D array of 12 random integers between 10 and 100.
# 3. Reshape the array into a 3x4 matrix.
# 4. Print the first 2 rows and the last 2 columns.
```

```python
import numpy as np

np.random.seed(42)

numbers = np.random.randint(10, 101, 12)
matrix = numbers.reshape(3, 4)

print("Original 3x4 Matrix:")
print(matrix)

sub_matrix = matrix[:2, -2:]

print("\nExtracted Sub-Matrix (First 2 rows, Last 2 columns):")
print(sub_matrix)
```

```python
# Task 4: Pandas DataFrame Filtering and Column Creation
#
# Problem Description:
# Create a DataFrame with Name, Department, Salary, and Experience.
# 1. Keep employees with Experience >= 3 and Salary > 6000.
# 2. Create Annual_Bonus using Salary * 0.10.
# 3. Sort the filtered DataFrame by Salary in descending order.
#
# Test Data:
# data = {
#     "Name": ["Ali", "Fatma", "Hassan", "Zainab", "Khaled"],
#     "Department": ["IT", "HR", "IT", "Finance", "IT"],
#     "Salary": [5000, 8000, 12000, 7500, 9000],
#     "Experience": [2, 4, 5, 3, 1]
# }
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Fatma", "Hassan", "Zainab", "Khaled"],
    "Department": ["IT", "HR", "IT", "Finance", "IT"],
    "Salary": [5000, 8000, 12000, 7500, 9000],
    "Experience": [2, 4, 5, 3, 1]
}

df = pd.DataFrame(data)

filtered_df = df[
    (df["Experience"] >= 3) & (df["Salary"] > 6000)
].copy()

filtered_df["Annual_Bonus"] = filtered_df["Salary"] * 0.10

filtered_df = filtered_df.sort_values(by="Salary", ascending=False)

print(filtered_df)
```
